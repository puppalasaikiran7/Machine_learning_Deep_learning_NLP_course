```python
#converting json to data frames

import pandas as pd 
from io import StringIO

data  = '{"employee_name":"saikiran", "email":"skrnpuppal@gmail.com", "job_profile": [{"title":"teamlead", "title2":"sr developer"}]}'
df = pd.read_json(StringIO(data))
```

```python
# reading the csv file using read_csv 
#converting the csv format to data frame

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data"
df1 = pd.read_csv(url,header=None)
df1
```

```python
# reading the tables from html pages 

url1 = "https://www.fdic.gov/bank-failures/failed-bank-list"
df4 = pd.read_html(url1)
df4
df4[0] 
```

```python
# if pandas request is having some issue then use the requests lib and use the headers ( mozilla/5.0) 

import requests 
from io import StringIO

# url3 = "https://www.fdic.gov/bank-failures/failed-bank-list"

url3 = "https://en.wikipedia.org/wiki/Mobile_country_code"

headers = {
    "user-agent" : "mozilla/5.0"
}


response = requests.get(url3,timeout=20,headers=headers)

print(response.status_code)

df6 = pd.read_html(StringIO(response.text),match="country")

df6[0]
```

```python
# reading the data from excel sheet 

# openpyxl is required for reading the excel sheet 

df7 = pd.read_excel('data.xlsx')
df7
```

```python
# pickles are usefull when you want to save the data permanently 
#when you close the python it forgets the data this pickle helps with that issue 


df7.to_pickle('pickle_df7.pkl')
```

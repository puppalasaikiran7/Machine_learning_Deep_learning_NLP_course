```python
import pandas as pd 

data = [1,2,3,4]
series = pd.Series(data) 
print(series)
```

```python
#create series from dictionaries 

data = {'a':1,'b':2,'c':3}

dict_series= pd.Series(data)
print(dict_series)
```

```python
# creating series using the index and data 

data = [1,2,3]
index_values = ['a','b','c']

indexed_series = pd.Series(data,index=index_values)
print(indexed_series)
```

```python
#creating data frame using dictionary of list 
data = {
    "name":["krish","john","klark"],
    "age":[22,34,45],
    "city":["banglore","new york","florida"]
}

df = pd.DataFrame(data)
print(df)
```

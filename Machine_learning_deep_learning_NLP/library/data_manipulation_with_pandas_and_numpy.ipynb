```python
# filling the missing values with the 0 using fillna 
df_filled = df.fillna(0)

df_filled.isnull().sum()
```

```python
#filling the missing value with the mean of the column 

df['sales_fillna']= df['sales'].fillna(df['sales'].mean())
df
```

```python
df.isnull().sum()
```

```python
#renaming the column names 
df = df.rename(columns={'date':'sales date'})
df.head()
```

```python
#finding the data type of the columns 

df.dtypes
```

```python
#changing the data type of the column 

df['sales'] =df['sales'].fillna(df['sales'].mean()).astype(int)
df.head()
```

```python
df.dtypes
```

```python
# apply method ( apply method lets you apply the function to your data ) 

df['new_value']=df['value'].apply(lambda x:x*2)
df
```

```python
# data aggregation and grouping 

df.head()
```

```python
# group by 

groupd_mean = df.groupby('product')['value'].mean()
print(groupd_mean)
```

```python
# group by ( multiple columns )
grouped_sum = df.groupby(['product','region'])['value'].sum()
print(grouped_sum)
```

```python
grouped_multiple_mean = df.groupby(['product','region'])['value'].mean()
print(grouped_multiple_mean)
```

```python
# aggregate method 
grouped_agg = df.groupby('region')['value'].agg(['mean','sum','count'])
print(grouped_agg)
```

```python
df['valuefillna']=df['value'].fillna(df['value'].mean())
df
```

```python
df.isnull().sum()
```

```python
df['revenue'] =df['sales_fillna']*df['valuefillna']
df
```

```python
df['revenue']= df['revenue'].astype(int)
df
```

```python
# multiple aggregation 
# .groupby(column).agg(variable_name = ("column","function"))

multiple_aggregations_groupby = df.groupby('region').agg(revenue_sum = ("revenue","sum") ,average_value =("valuefillna","mean"), units_sum = ("sales","sum") , orders=("sales","count"))
print(multiple_aggregations_groupby)
```

```python
# merging and joining data frames 

df1 = pd.DataFrame({'key':['a','b','c'] , 'values':[1,2,3]} )
df2 = pd.DataFrame({'key':['A','B','D'] , 'values':[4,5,6]} )
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1,df2,on='key',how='inner')
```

```python
pd.merge(df1,df2,on='key',how='outer')
```

```python
pd.merge(df1,df2,on='key',how='left')
```

```python
pd.merge(df1,df2,on='key',how='right')
```

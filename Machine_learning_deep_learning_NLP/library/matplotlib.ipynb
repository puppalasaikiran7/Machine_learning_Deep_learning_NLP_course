##### Line plot ( how the values changes over a continuous sequence , over the time )[stock price , company stock over time , temperature over the months ]

##### Bar plot (compares values across different categorical )[sales of different product ,population of different country ]

##### Histogram ( numerical data is distributed by grouping into intervals called bins )

##### Scatter plot ( relationship between 2 numerical variables using individual points )

##### Pie plot ( how individual categories contribute to whole )

```python
import matplotlib.pyplot as plt 

x = [1,2,3,4]
y = [5,6,7,8]

#create a line plot 

plt.plot(x,y)
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.title("line plotting")
plt.show()
```

```python
# color , linestyle and marker for the plotting

plt.plot(x,y,color="red",linestyle="--",marker=".")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("double dash plots")
plt.show()
```

```python
# markersize and linewidth 

plt.plot(x,y,color="green",linestyle="--",marker="*",linewidth=3,markersize=15)
```

```python
# grid 

plt.plot(x,y,color="darkgreen",linestyle='--',marker='o',linewidth=5,markersize=10)
plt.grid(True)
```

```python
#subplots 

x  = [1,2,3,4]
y1 = [5,6,7,8]
y3 = [9,10,11,12]

# figure ( creates the blank canvas )
plt.figure(figsize=(9,5))

plt.subplot(2,2,1)
plt.plot(x,y1,color="green",linestyle='--',linewidth=10,marker="*",markersize=6)
plt.title("sub plots 1")
plt.show()

plt.subplot(2,2,2)
plt.plot(y1,y3,color="red",linestyle='--',linewidth=5,marker='*',markersize=6)
plt.title("sub plot 2")
plt.show()

plt.subplot(2,2,3)
plt.plot(y1,x,color="purple",linestyle="-.",linewidth=5,marker="*",markersize=20,markerfacecolor = "green")
plt.grid(True)
plt.title("sub plot 3")
plt.show()

plt.subplot(2,2,4)
plt.plot(y3,x,color="blue",linestyle="-.",linewidth=5,marker="*",markersize=5,markerfacecolor="darkgreen")
plt.title("sub plot 4")
plt.show()
```

```python
#bar plot ( categorical purpose like a has this many values , b has this many values )

categories = ['a','b','c','d']
values = [5,6,3,5]

plt.bar(categories,values,color="green")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("bar plot")
plt.grid(True)
```

```python
# Histogram ( numerical values )

import numpy as np

data = np.random.randint(40,100,100)

# alpha ( transparency between 0 to 1 ) 
# label is used for the legend purpose ( top left icon )
# width of bin ( max value - min value / total number of bins )

plt.hist(data,bins=10,edgecolor="black",color="blue",linewidth=1.2,alpha=0.5,label="student data")
plt.title("histgram plotting")
plt.xlabel("marks")
plt.ylabel("number of students")
plt.grid(True,alpha=0.6)
plt.legend()
plt.show()
```

```python
#scatter plot 

x = [1,2,3,4]
y = [2,3,4,5]

plt.scatter(x,y,color="green",marker="X")
```

```python
# pie chart 

labels = ['a','b','c','d']
sizes = [30,20,40,10]
colors = ['gold','yellowgreen','lightcoral','lightskyblue']
explode = (0.2,0,0,0)

plt.pie(sizes , labels= labels, colors=colors,autopct="%1.1f%%",shadow=True, explode = explode)
```

```python
import matplotlib.pyplot as plt 

x = [1,2,3,4]
y = [5,6,7,8]

#create a line plot 

plt.plot(x,y)
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.title("line plotting")
plt.show()
```

```python
# color , linestyle and marker for the plotting

plt.plot(x,y,color="red",linestyle="--",marker=".")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("double dash plots")
plt.show()
```

```python
# markersize and linewidth 

plt.plot(x,y,color="green",linestyle="--",marker="*",linewidth=3,markersize=15)
```

```python
# grid 

plt.plot(x,y,color="darkgreen",linestyle='--',marker='o',linewidth=5,markersize=10)
plt.grid(True)
```

```python
#subplots 

x  = [1,2,3,4]
y1 = [5,6,7,8]
y3 = [9,10,11,12]

# figure ( creates the blank canvas )
plt.figure(figsize=(15,15))

plt.subplot(1,2,1)
plt.plot(x,y1,color="green",linestyle='--',linewidth=10,marker="*",markersize=6)
plt.title("sub plots 1")
plt.show()

plt.subplot(1,2,2)
plt.plot(y1,y3,color="red",linestyle='--',linewidth=15,marker='*',markersize=6)
plt.title("sub plot 2")
plt.show()
```

```python
import random

empty_list = []
non_duplicate_list = []

for i in range(10):
    random_value = random.randint(1,10)
    empty_list.append(random_value)

#sorting the list 
empty_list.sort()
print(empty_list)

#reverse the list 
empty_list.reverse()
print(empty_list)

#removing the duplicates 
for i in empty_list:
    if i not in non_duplicate_list:
        non_duplicate_list.append(i)
        
print(non_duplicate_list)
```

```python
#13th solution

def reverse_list(lst):
  return sorted(lst , reverse=True)

original_lst = [1,2,6,7,3,7,8,9,3,7,3,0,1]

reversed_list = reverse_list(original_lst)
print(reversed_list)
print(original_lst)
```

```python
#14th solution ( rotated list )

def rotated_lst(lst,n):
  return lst[n:]+lst[:n]

lst = [1,2,3,4,5]
n = int(input("enter the nth position: "))
rotated = rotated_lst(lst,n)
print(rotated)

```

```python
#15th solution  ( intersection list )

def intersection(lst1 , lst2):
  return [x for x in lst1 if x in lst2]

lst1 = [1,2,3,4,5]
lst2 =[3,4,5,6,7]
intersection_value = intersection(lst1 , lst2)
print(intersection_value)
```

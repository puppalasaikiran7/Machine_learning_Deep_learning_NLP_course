```python
def division(value1,value2):
    try:
        final_value = value1/value2
        
    except Exception as ex:
        print(ex)
        final_value = None 
    finally:
        print(f"this is done")
    return final_value
        

print(division(3,0))
```

```python

def file_reading(file_name):
    try:
        with open(file_name,'r')as file:
            reading = file.read()
            return reading 
    except Exception as ex:
        print(ex)
    finally:
        print("this is done")

print(file_reading('data.txt'))
```

```python
def sum_of_list(lst):
    total = 0
    try:
        for i in lst:
            if not isinstance(i,int):
                return TypeError(f"{i} is not an integer")
            else:
                total+=i
        return total
    except TypeError as ex:
        print(f"typeerror: {ex}")
    except Exception as ex:
        print(ex)
    finally:
        print("this is done")



print(sum_of_list([0,1,2,3]))
```

```python
def user_input():
    value = None
    try:
        value = int(input("enter the value: "))

    except ValueError as ex:
        print(f"valueerror: {ex}")

    finally:
        print("execution completed")
    return value

print(user_input())
```

```python
def value_of_key(dic,key):
    try:
        value = dic[key]
    except KeyError as ex:
        print(f"key error",ex)
        value = None
    finally:
        print("this is  done")

    return value

print(value_of_key({'a':1,'b':2,'c':3},'d'))
```

```python
def list_operation(lst,index):
    try:
        value = lst[index]
    except IndexError as ex:
        print(f"index error for {ex}")
        value = None
    finally:
        print("this is done")
    return value

print(list_operation([1,2,3,4,5],9))
```

```python
import requests

def read_url(url):
    try:
        response = requests.get(url,timeout=10)
        response.raise_for_status()
        print(response.status_code)
        print(response.reason)
        print(response.headers)
        return response.text
    
    except requests.RequestException as ex:
        print("network error: ",ex)
    finally:
        print("this is done")
    

# print(read_url('https://jsonplaceholder.typicode.com/posts/1'))
# print(read_url('https://nonexistent.url'))
```

```python
#creating the custom exception 

class negative_number_error(Exception):
    pass

def custom_exception(lst):
    try:
        for i in lst:
            if i<0:
                raise negative_number_error(f"there is a negative number {i}")
    except negative_number_error as ex:
        print(ex)
    finally:
        print("this is  done")


custom_exception([1,2,3,4,-5])
```

```python
class Calculator:
    def __init__(self,a,b):
        self.a = a 
        self.b = b 
    def divide(self):
        try:
            result = self.a/self.b 
        except Exception as ex:
            print(ex)
            result = None
        finally:
            print("this is done")
        return result
    
calc = Calculator(10,0)
print(calc.divide())
```

```python
def convert_to_integer(lst):
    integers = []
    try:
        for i in lst:
            integers.append(int(i))
    except ValueError as ex:
        print(ex)
        integers = None
    finally:
        print("this is done")
    return integers


# print(convert_to_integer(["1","2","3","4"]))
print(convert_to_integer(['1', '2', 'three', '4']))
```

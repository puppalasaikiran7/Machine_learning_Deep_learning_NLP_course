```python
import sqlite3

def database_connection():
    
    connection = sqlite3.connect("test.db")

    cursor = connection.cursor()

    cursor.execute(
        """ 
    create table if not exists employees (
    id integer,
    name text ,
    age integer,
    department text
    )
    """
    )

    connection.commit()
    connection.close()

database_connection()
```

```python
def inserting_value(id,name,age,department):

    connection = sqlite3.connect("test.db")
    
    cursor = connection.cursor()
    
    cursor.execute(
        """ 
    insert or ignore into employees (id ,name , age , department) values (?,?,?,?)
    """,(id,name,age,department)
    )

    connection.commit()
    connection.close()
    print("employee successully created")


inserting_value(1,"saikiran",23,"AIML & robotics research")
```

```python
inserting_value(2,"james",25,"marketing")
inserting_value(3,"harvey",30,"robotics")
inserting_value(4,"patrick",25,'aiml')
```

```python
def fetch_all_data():
    connection = sqlite3.connect('test.db')
    cursor = connection.cursor()

    cursor.execute("select count(*) from employees")
    print(cursor.fetchone())

    cursor.execute(
        """ 
            select * from employees
        """
    )
    records = cursor.fetchall()
    print(records)
    connection.close()
    for record in records:
        print(record)

fetch_all_data()
```

```python
import os

def writing_file(output_file,list_value):
    if os.path.exists(output_file):
        print("file exist")
    else:
        with open(output_file,'w') as file:
            for i in list_value:
                file.write(i+'\n')

def reading(output_file):
    with open(output_file,'r')as file:
        print(file.read())


list_value = ["hello","world"]
writing_file("output_file.txt",list_value)
reading("output_file.txt")
```

```python
def appending(file_name,text):
    with open(file_name,'a')as file:
        file.write(text+'\n')

appending('log.txt',"this is the first log")
```

```python
def counting(file_name):
    
    with open(file_name,'r')as file:
        string_value = file.read()
        list_value = string_value.split()
        return len(list_value)

counting("log.txt")
```

```python
def replacing(file_name):
    with open(file_name,'r')as file:
        string_value = file.read()

    replaced_string = string_value.replace('[COMPANY_NAME]','google')

    with open(file_name,'w')as file:
        file.write(replaced_string)

replacing('data.txt')
```

```python
def reversereading(files):
    with open(files,'r')as file:
        strings = file.read()
        reverse_string = strings[::-1]
        print(reverse_string)

reversereading('reverse.txt')
```

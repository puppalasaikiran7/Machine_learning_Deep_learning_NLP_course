```python
#1st solution
set_value = set(range(1,11))
print(set_value)
```

```python
#2nd solution
set_value.add(11)
print(set_value)
```

```python
#3rd solution
first_set = set(range(1,6))
second_set = set(range(2,11,2))

union_value = first_set.union(second_set)
intersection_value = first_set.intersection(second_set)
difference_value = first_set.difference(second_set)
symmetric_value = first_set.symmetric_difference(second_set)

print(union_value)
print(intersection_value)
print(difference_value)
print(symmetric_value)
```

```python
#7th solution

first_set_number = set(range(1,6))
second_set_number = set(range(1,4))

subset = first_set_number.issubset(second_set_number)
print(subset)

superset = first_set_number.issuperset(second_set_number)
print(superset)
```

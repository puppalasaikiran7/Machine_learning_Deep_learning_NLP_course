```python
#15th solution


def min_value(tup):
    return min(tup)

def max_value(tup):
    return max(tup)

def sum_value(tup):   # fixed spelling here
    return sum(tup)

tup = (1, 2, 3, 4, 5)
print(min_value(tup))   # 1
print(max_value(tup))   # 5
print(sum_value(tup))   # 15
```

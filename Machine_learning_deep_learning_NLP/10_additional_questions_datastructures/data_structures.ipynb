#### sum of values in the list

```python
def sum_list(numbers):
    # Your code goes here
    total =0
    for i in numbers:
        total+=i
    return total

numbers = [1, 2, 3, 4, 5]
sum_list(numbers)
```

#### find the largest value in the list

```python
def find_largest(numbers):
    # Your code goes here
    return max(numbers)

numbers = [3, 8, 2, 10, 5]
print(find_largest(numbers))
```

#### Remove duplicate from the list

```python
def remove_duplicates(lst):
    # Your code goes here
    output_list=[]
    for i in lst:
        if i in output_list:
            continue
        else:
            output_list.append(i)
    return output_list

lst = [1, 2, 2, 3, 4, 4, 5]
print(remove_duplicates(lst))
```

#### check if the values in the list are unique

```python
def check_unique(lst):
    # Your code goes here
    if len(lst)==len(set(lst)):
        return True

lst = [1, 2, 3, 4, 5]
check_unique(lst)
```

#### reverse the list no slicing approach 

```python
def reverse_list(lst):
    return sorted(lst,reverse = True)


lst = [1,2,3,4,5]
reverse_list(lst)
```

#### number of even elements or odd elements in list 

```python
def count_even_odd(lst):
    # Your code goes here
    even_count = 0
    odd_count = 0
    
    for i in lst:
        if i%2==0:
            even_count+=1
        else:
            odd_count+=1
    tuple_value = (even_count,odd_count)
    return tuple_value


lst = [1,2,3,4,5]
count_even_odd(lst)
```

#### check if one is sublist of another list 


```python
def check_sub_list(lst1,lst2):
    for i in lst1:
        found = False
        for j in lst2:
            if i==j:
                found = True
                break
        if not found:
            return False
    return True


lst1 = [1, 2, 3]
lst2 = [1, 2, 3, 4, 5]

print(check_sub_list(lst1,lst2))
```

#### max difference between two consecutive number 

```python
def max_difference(lst):
    max_difference = 0 
    for i in range(len(lst)-1):
        difference = abs(lst[i]-lst[i+1])
        if difference>max_difference:
            max_difference = difference
    return max_difference


lst = [1, 7, 3, 10, 5]
max_difference(lst)
```

#### merge two sorted list 

```python
def sorted_merge(list1,list2):
    list1.sort()
    list2.sort()
    merged_list = sorted(list1+list2)
    return merged_list

list1 = [1, 3, 5]
list2 = [2, 4, 6]
sorted_merge(list1,list2)
```

#### rotate a list with out slicing

```python
def rotate(lst,k):
    if k==0:
        return lst
    if len(lst)==0:
        return lst
    if len(lst)==1:
        return lst

    #if the k values crosses the len of list condition

    n=len(lst)
    k = k%n

    for i in range(k):
        last_value = lst.pop()
        lst.insert(0,last_value)
    
    return lst

lst = [1, 2, 3, 4, 5]
k = 2 # number of positions to rotate the list 
rotate(lst,k)
```

#### merge 2 list into dictionary 

```python
def merge_list_dict(keys,values):
    if len(keys)!=len(values):
        return False
    return dict(zip(keys,values))


keys = ['a', 'b', 'c']
values = [1, 2, 3]
merge_list_dict(keys,values)
```

#### merge multiple dictionaries 

```python
def merge_three_dictionaries(dict1,dict2,dict3):
    dict1.update(dict2)
    dict1.update(dict3)
    return dict1

print(merge_three_dictionaries({'a': 1, 'b': 2}, {'c': 3, 'd': 4}, {'e': 5, 'f': 6}))
```

#### word frequency in sentence 

```python
def count_word_frequency(sentence):
    dictio ={}
    list_value = sentence.split()
    for i in list_value:
        if i in dictio:
            dictio[i]+=1
        else:
            dictio[i]=1
    return dictio

print(count_word_frequency("hello world hello"))
```

#### palindrome tuple

```python
def is_palonrome(tup):
    if tup[:]==tup[::-1]:
        return True
    else:
        return False

print(is_palonrome((1, 2, 3, 2, 1)))
```

#### merge two dictionaries with common keys 

```python
def common_key_values_in_dict(dicts):
    dics = {}
    for i in dicts:
        for keys,values in i.items():
            if keys in dics:
                dics[keys]+=values
            else:
                dics[keys]=values
    return dics

print(common_key_values_in_dict([{'a': 1, 'b': 2}, {'b': 3, 'c': 4}, {'c': 5, 'd': 6}]))

```

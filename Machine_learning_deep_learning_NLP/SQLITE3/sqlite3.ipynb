```python
import sqlite3 
```

```python
#creating the connection 

connection = sqlite3.connect("practice.db")
```

```python
# creating the cursor ( for executing the sql commands )

cursor = connection.cursor()
```

```python
#creating the table 
cursor.execute(
    """
create table if not exists employees(
id int primary key,
name varchar(20) not null,
age int,
department varchar(30)
)
 """
)
```

```python
# commiting the changes 
connection.commit()
```

```python
# table list ( show table ) ( pragma table_list )
cursor.execute("PRAGMA table_list;")
tables = cursor.fetchall()

tables
```

```python
# inserting the data in the tables 

cursor.execute(
    """
insert or ignore into employees(id,name,age,department) values (1,"saikiran",23,"AI"),(2,"vishnu",24,"IT"),(3,"pavan",32,"marketing")
"""
)
```

```python
connection.commit()
```

```python
# fetching the values from the table

cursor.execute("select * from employees")
result = cursor.fetchall()
print(result)
```

```python
# updating the employees 

cursor.execute(
    """ 
    update employees set age = 34 where name="pavan"
"""
)


connection.commit()
```

#### working with sales data 

```python
connection1 = sqlite3.connect("sales_data.db")

cursor1 = connection1.cursor()

cursor1.execute(
    """ 
create table if not exists sales(
id INTEGER primary key autoincrement ,
date data not null,
product varchar(50),
sales decimal(10,1),
region varchar(40)
)
"""
)

sales_data = [
    ("2023-01-20","pen",1000,"hyderabad"),
    ("2023-04-30","glass bottles",50000,"janagam"),
    ("2023-07-10","sweaters",200,"gachibowli")
]

cursor1.executemany(
    """ 
insert or ignore into sales (date , product,sales , region ) values (?,?,?,?) 
""",sales_data)

connection1.commit()
```

```python
cursor1.execute("select * from sales")

result4  = cursor1.fetchall()
print(result4)
```

```python
connection1.close()
```

```python
connection3 = sqlite3.connect("inventory.db")
```

```python
cursor4 = connection3.cursor()
```

```python
cursor4.execute(
    """ 
create table if not exists products (
id int primary key,
name varchar(40) not null,
price int ,
quantity int
)
"""
)

connection3.commit()
```

```python
products = [
    (1,"laptop",55000,10),
    (2,"mouse",500,50),
    (3,"keyboard",1200,30),
    (4,"monitor",8000,15)
]

cursor4.executemany(
    """ 
insert into products (id , name , price , quantity) values(?,?,?,?)
"""
,products)

```

```python
cursor4.execute("select * from products")

result5 = cursor4.fetchall()

for result in result5:
    print(result)
```

```python
cursor4.execute(
    """ 
update products set quantity=45 where name="mouse"
"""
)

connection3.commit()
```

```python
cursor4.execute(
    """ 
delete from products where name="keyboard"
"""
)

connection3.commit()

connection3.close()
```

```python
#inventory example with single inheritance

class inventory():
    def __init__(self,shop_name):
        self.shop_name = shop_name 
        self.items = []  # this is the correct way to add the empty list in the constructor 

    def add_item(self,*item):
        self.items.extend(item)
        return f"item {item} add to inventory"
    
    def get_item_count(self):
        return len(self.items)
    

class advance_inventory(inventory):
    def __init__(self,shop_name , location):
        inventory.__init__(self,shop_name)
        self.location = location 
        self.item_price = {}
    
    def add_item_with_price(self,item,price):
        self.items.append(item)
        self.item_price[item] = price
        return f"added {item} at {price}"
    
    def total_value(self):
        total = sum(self.item_price.values())
        return total
    
    def get_most_expensive_item(self):
        max_price = max(self.item_price.values())
        expensive_item = [item for item,price in self.item_price.items() if price==max_price] 
        return f"{expensive_item} at {max_price}"
    
    def search_item(self,item_name):
        if  item_name in self.item_price:
            return f"{item_name} is available in inventory"
        else:
            return f"{item_name} is not available in inventory"

store= advance_inventory("TechStore","Downtown")
print(store.add_item_with_price("laptop",50000))
print(store.add_item_with_price("mouse",3000))
print(store.add_item_with_price("keyboard",10000))
print(store.add_item_with_price("monitor",20000))

print(store.get_item_count())
print(store.total_value())
print(store.get_most_expensive_item())
item_name = input("enter the material to check: ")
print(store.search_item(item_name))
```

### heirarchical inheritance 

```python
class bankaccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def __str__(self):
        return f"Account Owner: {self.owner} | Balance: ₹{self.balance}"

    def deposit(self, deposite_amount):
        try:
            if not isinstance(deposite_amount, (int, float)):
                raise Exception("amount should be a number")
            if deposite_amount <= 0:
                raise Exception("amount should be positive")
            self.balance += deposite_amount
            return f"Deposited ₹{deposite_amount}. Balance: ₹{self.balance}"
        except Exception as e:
            return f"Error: {e}"

    def withdrawal(self, withdraw_amount):
        try:
            if not isinstance(withdraw_amount, (int, float)):
                raise Exception("amount should be a number")
            if withdraw_amount <= 0:
                raise Exception("amount should be positive")
            if withdraw_amount > self.balance:
                raise Exception("insufficient balance")
            self.balance -= withdraw_amount
            return f"Withdrawn ₹{withdraw_amount}. Balance: ₹{self.balance}"
        except Exception as e:
            return f"Error: {e}"


class saving_account(bankaccount):
    def __init__(self, owner, balance=0, interest=0.0025):
        super().__init__(owner, balance)
        self.interest = interest

    def __str__(self):
        return f"Savings Account | Owner: {self.owner} | Balance: ₹{self.balance} | Interest Rate: {self.interest*100}%"

    def add_interest(self):
        interest_amount = self.balance * self.interest
        self.balance += interest_amount
        return f"Interest added: ₹{interest_amount}. New balance: ₹{self.balance}"


class current_account(bankaccount):
    def __init__(self, owner, balance=0, overdraft_limit=10000,
                 overdraft_interest=0.02, transaction_fee=10,
                 free_transaction_count_per_month=45, monthly_charge=0.02):
        super().__init__(owner, balance)
        self.overdraft_limit = overdraft_limit
        self.overdraft_interest = overdraft_interest
        self.transaction_fee = transaction_fee
        self.transaction_count = 0
        self.free_transaction_count_per_month = free_transaction_count_per_month
        self.monthly_charge = monthly_charge

    def __str__(self):
        return (f"Current Account | Owner: {self.owner} | Balance: ₹{self.balance} | "
                f"Overdraft Limit: ₹{self.overdraft_limit} | "
                f"Transactions: {self.transaction_count}/{self.free_transaction_count_per_month}")

    def deposit(self, deposit_amount):
        result = super().deposit(deposit_amount)  # ✅ correct method name
        if not result.startswith("Error"):
            self.transaction_count += 1
            return f"Deposited ₹{deposit_amount}. Balance: ₹{self.balance}"  # ✅ fixed variable
        return result

    def withdrawal(self, withdraw_amount):
        try:
            if not isinstance(withdraw_amount, (int, float)):
                raise Exception("withdraw amount should be a number")
            if withdraw_amount <= 0:
                raise Exception("withdraw amount should be positive")
            if withdraw_amount > self.balance + self.overdraft_limit:
                raise Exception("exceeds overdraft limit")
            self.balance -= withdraw_amount
            self.transaction_count += 1
            return f"Withdrawn ₹{withdraw_amount}. Balance: ₹{self.balance}"
        except Exception as e:
            return f"Error: {e}"

    def monthly_charges(self):
        print(f"\n--- Monthly Charges for {self.owner} ---")

        # 1. monthly fee as percentage of balance
        monthly_charge_amount = self.balance * self.monthly_charge
        if self.balance - monthly_charge_amount < -self.overdraft_limit:
            print("Monthly charge exceeds overdraft limit!")
        else:
            self.balance -= monthly_charge_amount
            print(f"Monthly fee: ₹{monthly_charge_amount} ({self.monthly_charge*100}%). Balance: ₹{self.balance}")

        # 2. overdraft interest — use print not return!
        if self.balance < 0:
            overdraft_interest_amount = abs(self.balance) * self.overdraft_interest  # ✅ self.
            self.balance -= overdraft_interest_amount  # ✅ fixed typo
            print(f"Overdraft interest: ₹{overdraft_interest_amount}. Balance: ₹{self.balance}")
        else:
            print("No overdraft interest.")

        # 3. transaction charges — now actually reachable!
        if self.transaction_count > self.free_transaction_count_per_month:  # ✅ self.
            extra = self.transaction_count - self.free_transaction_count_per_month  # ✅ fixed order
            charges = extra * self.transaction_fee
            self.balance -= charges
            print(f"Extra transaction charges ({extra} transactions): ₹{charges}. Balance: ₹{self.balance}")
        else:
            print(f"No extra transaction charges. ({self.transaction_count}/{self.free_transaction_count_per_month} used)")

        self.transaction_count = 0  # reset for next month
        print(f"--- Final Balance: ₹{self.balance} ---\n")


print("=" * 60)
print("SAVINGS ACCOUNT TESTS")
print("=" * 60)

# Object creation
sai = saving_account("Saikiran", balance=5000)
print(sai)

print("\n--- Deposit Tests ---")
print(sai.deposit(1000))        # ✅ normal deposit
print(sai.deposit(-500))        # ❌ negative amount
print(sai.deposit(0))           # ❌ zero amount
print(sai.deposit("abc"))       # ❌ string instead of number
print(sai.deposit(None))        # ❌ None value

print("\n--- Withdrawal Tests ---")
print(sai.withdrawal(500))      # ✅ normal withdrawal
print(sai.withdrawal(-100))     # ❌ negative amount
print(sai.withdrawal(0))        # ❌ zero amount
print(sai.withdrawal("abc"))    # ❌ string instead of number
print(sai.withdrawal(99999))    # ❌ insufficient balance

print("\n--- Interest Tests ---")
print(sai.add_interest())       # ✅ normal interest
sai2 = saving_account("Ravi", balance=0)
print(sai2.add_interest())      # ⚠️ interest on zero balance

print("\n--- __str__ Test ---")
print(sai)                      # ✅ shows updated balance


print("=" * 60)
print("CURRENT ACCOUNT TESTS")
print("=" * 60)

# Object creation
ram = current_account("Ramesh", balance=15000)
print(ram)

print("\n--- Deposit Tests ---")
print(ram.deposit(2000))       # ✅ normal deposit
print(ram.deposit(-500))       # ❌ negative amount
print(ram.deposit(0))          # ❌ zero amount
print(ram.deposit("abc"))      # ❌ string instead of number
print(ram.deposit(None))       # ❌ None value

print("\n--- Withdrawal Tests ---")
print(ram.withdrawal(5000))      # ✅ normal withdrawal
print(ram.withdrawal(-100))      # ❌ negative amount
print(ram.withdrawal(0))         # ❌ zero amount
print(ram.withdrawal("abc"))     # ❌ string instead of number
print(ram.withdrawal(99999))     # ❌ exceeds overdraft limit

print("\n--- Overdraft Tests ---")
ram2 = current_account("Ramesh2", balance=500)
print(ram2.withdrawal(5000))     # ✅ overdraft kicks in — balance goes negative
print(ram2)                     # shows negative balance
print(ram2.withdrawal(6000))     # ❌ exceeds overdraft limit (500 + 10000 = 10500 but already used some)

print("\n--- Transaction Count Tests ---")
ram3 = current_account("Ramesh3", balance=100000, free_transaction_count_per_month=3)  # low limit for testing
print(ram3.deposit(100))       # ✅ free transaction 1
print(ram3.deposit(100))       # ✅ free transaction 2
print(ram3.deposit(100))       # ✅ free transaction 3
print(ram3.deposit(100))       # ⚠️ extra transaction — fee charged
print(ram3.deposit(100))       # ⚠️ extra transaction — fee charged
print(ram3)

print("\n--- Monthly Charges Tests ---")

# case 1 — normal positive balance
ram4 = current_account("Ramesh4", balance=15000)
print(ram4.deposit(1000))
print(ram4.withdrawal(2000))
ram4.monthly_charges()          # ✅ normal monthly charges
print(ram4)

# case 2 — negative balance (overdraft interest applies)
ram5 = current_account("Ramesh5", balance=500)
print(ram5.withdrawal(8000))     # balance goes negative
ram5.monthly_charges()          # ✅ overdraft interest should apply
print(ram5)

# case 3 — excess transactions + monthly charges
ram6 = current_account("Ramesh6", balance=50000, free_transaction_count_per_month=2)
print(ram6.deposit(100))       # free
print(ram6.deposit(100))       # free
print(ram6.deposit(100))       # extra — fee charged
print(ram6.withdrawl(100))      # extra — fee charged
ram6.monthly_charges()          # transaction charges in monthly too
print(ram6)

# case 4 — zero balance monthly charges
ram7 = current_account("Ramesh7", balance=0)
ram7.monthly_charges()          # ⚠️ monthly charge on zero balance
print(ram7)
```

```python
#example 1 

class Animal:
    def speak(self):
        return "sound of the animal"

class Dog(Animal):
    def speak(self):
        return "woof woof"

class Cat(Animal):
    def speak(self):
        return "meow meow"
    

dog = Dog()
cat = Cat()
print(dog.speak()) 
print(cat.speak())
```

```python
#example 2 
class Shape:
    def area(self):
        return "the area of the figure"

class Rectangle(Shape):
    def __init__(self,height,width):
        self.height = height
        self.width = width 
    
    def area(self):
        return self.height*self.width

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    
    def area(self):
        return 3.14*self.radius*self.radius

def print_area(shape):
    return f"the area is {shape.area()}"

rectangle = Rectangle(10,20)
circle=Circle(10)

print_area(circle)

```

```python
#example 3 

class Employee:
    def __init__(self, name, emp_id, salary):
        self.name = name
        self.emp_id = emp_id
        self.salary = salary
    
    def display_info(self):
        print(f"Name: {self.name}")
        print(f"ID: {self.emp_id}")
        print(f"Salary: ${self.salary}")
    
    def get_annual_bonus(self):
        return self.salary * 0.10  # 10% bonus

class Developer(Employee):
    def __init__(self, name, emp_id, salary, programming_languages):
        super().__init__(name, emp_id, salary)
        self.programming_languages = programming_languages
    
    def display_info(self):
        super().display_info()
        print(f"Languages: {', '.join(self.programming_languages)}") 
    
    def get_annual_bonus(self):
        # Developers get 15% bonus
        return self.salary * 0.15

class Manager(Employee):
    def __init__(self, name, emp_id, salary, team_size):
        super().__init__(name, emp_id, salary)
        self.team_size = team_size
    
    def display_info(self):
        super().display_info()
        print(f"Team Size: {self.team_size}")
    
    def get_annual_bonus(self):
        # Managers get 20% bonus + $1000 per team member
        base_bonus = self.salary * 0.20
        team_bonus = self.team_size * 1000
        return base_bonus + team_bonus

class Intern(Employee):
    def __init__(self, name, emp_id, salary, university):
        super().__init__(name, emp_id, salary)
        self.university = university
    
    def display_info(self):
        super().display_info()
        print(f"University: {self.university}")
    
    def get_annual_bonus(self):
        # Interns get 5% bonus
        return self.salary * 0.05

# Create employees
dev = Developer("Saikiran", "D001", 80000, ["Python", "JavaScript", "Java"])
manager = Manager("Saikiran", "M001", 100000, 10)
intern = Intern("Saikiran", "I001", 30000, "MIT")

# Display info
print("=== Developer ===")
dev.display_info()
print(f"Annual Bonus: ${dev.get_annual_bonus():.2f}\n")

print("=== Manager ===")
manager.display_info()
print(f"Annual Bonus: ${manager.get_annual_bonus():.2f}\n")

print("=== Intern ===")
intern.display_info()
print(f"Annual Bonus: ${intern.get_annual_bonus():.2f}")


```

```python
class bankaccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def __str__(self):
        return f"Account Owner: {self.owner} | Balance: ₹{self.balance}"

    def deposit(self, deposite_amount):
        try:
            if not isinstance(deposite_amount, (int, float)):
                raise Exception("amount should be a number")
            if deposite_amount <= 0:
                raise Exception("amount should be positive")
            self.balance += deposite_amount
            return f"Deposited ₹{deposite_amount}. Balance: ₹{self.balance}"
        except Exception as e:
            return f"Error: {e}"

    def withdrawal(self, withdraw_amount):
        try:
            if not isinstance(withdraw_amount, (int, float)):
                raise Exception("amount should be a number")
            if withdraw_amount <= 0:
                raise Exception("amount should be positive")
            if withdraw_amount > self.balance:
                raise Exception("insufficient balance")
            self.balance -= withdraw_amount
            return f"Withdrawn ₹{withdraw_amount}. Balance: ₹{self.balance}"
        except Exception as e:
            return f"Error: {e}"


class saving_account(bankaccount):
    def __init__(self, owner, balance=0, interest=0.0025):
        super().__init__(owner, balance)
        self.interest = interest

    def __str__(self):
        return f"Savings Account | Owner: {self.owner} | Balance: ₹{self.balance} | Interest Rate: {self.interest*100}%"

    def add_interest(self): 
        interest_amount = self.balance * self.interest
        self.balance += interest_amount
        return f"Interest added: ₹{interest_amount}. New balance: ₹{self.balance}"


class current_account(bankaccount):
    def __init__(self, owner, balance=0, overdraft_limit=10000,
                 overdraft_interest=0.02, transaction_fee=10,
                 free_transaction_count_per_month=45, monthly_charge=0.02):
        super().__init__(owner, balance)
        self.overdraft_limit = overdraft_limit
        self.overdraft_interest = overdraft_interest
        self.transaction_fee = transaction_fee
        self.transaction_count = 0
        self.free_transaction_count_per_month = free_transaction_count_per_month
        self.monthly_charge = monthly_charge

    def __str__(self):
        return (f"Current Account | Owner: {self.owner} | Balance: ₹{self.balance} | "
                f"Overdraft Limit: ₹{self.overdraft_limit} | "
                f"Transactions: {self.transaction_count}/{self.free_transaction_count_per_month}")

    def deposit(self, deposite_amount):
        result = super().deposit(deposite_amount)  # ✅ correct method name
        if not result.startswith("Error"):
            self.transaction_count += 1
            return f"Deposited ₹{deposite_amount}. Balance: ₹{self.balance}"  # ✅ fixed variable
        return result

    def withdrawal(self, withdraw_amount):
        try:
            if not isinstance(withdraw_amount, (int, float)):
                raise Exception("withdraw amount should be a number")
            if withdraw_amount <= 0:
                raise Exception("withdraw amount should be positive")
            if withdraw_amount > self.balance + self.overdraft_limit:
                raise Exception("exceeds overdraft limit")
            self.balance -= withdraw_amount
            self.transaction_count += 1
            return f"Withdrawn ₹{withdraw_amount}. Balance: ₹{self.balance}"
        except Exception as e:
            return f"Error: {e}"

    def monthly_charges(self):
        print(f"\n--- Monthly Charges for {self.owner} ---")

        # 1. monthly fee as percentage of balance
        monthly_charge_amount = self.balance * self.monthly_charge
        if self.balance - monthly_charge_amount < -self.overdraft_limit:
            print("Monthly charge exceeds overdraft limit!")
        else:
            self.balance -= monthly_charge_amount
            print(f"Monthly fee: ₹{monthly_charge_amount} ({self.monthly_charge*100}%). Balance: ₹{self.balance}")

        # 2. overdraft interest — use print not return!
        if self.balance < 0:
            overdraft_interest_amount = abs(self.balance) * self.overdraft_interest  # ✅ self.
            self.balance -= overdraft_interest_amount  # ✅ fixed typo
            print(f"Overdraft interest: ₹{overdraft_interest_amount}. Balance: ₹{self.balance}")
        else:
            print("No overdraft interest.")

        # 3. transaction charges — now actually reachable!
        if self.transaction_count > self.free_transaction_count_per_month:  # ✅ self.
            extra = self.transaction_count - self.free_transaction_count_per_month  # ✅ fixed order
            charges = extra * self.transaction_fee
            self.balance -= charges
            print(f"Extra transaction charges ({extra} transactions): ₹{charges}. Balance: ₹{self.balance}")
        else:
            print(f"No extra transaction charges. ({self.transaction_count}/{self.free_transaction_count_per_month} used)")

        self.transaction_count = 0  # reset for next month
        print(f"--- Final Balance: ₹{self.balance} ---\n")


# ✅ Objects
sai = saving_account("Saikiran", balance=5000)
print(sai)                    # __str__
print(sai.deposit(1000))
print(sai.add_interest())

ram = current_account("ramesh", balance=15000)
print(ram)                    # __str__
print(ram.deposit(2000))
print(ram.withdrawal(5000))
ram.monthly_charges()
print(ram)                    # __str__ after charges

```

### Abstraction

```python
#polymorphism with abstraction 

from abc import ABC,abstractmethod 

class Shape(ABC):       
    # init constructor is allowed 
    
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length 
        self.width = width 

    def area(self):
        return self.length * self.width 

    def perimeter(self):
        return 2*(self.length+self.width)


class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    
    def area(self):
        return 3.14*self.radius*self.radius
    
    def perimeter(self):
        return 2*3.14*self.radius
   
    
def calculate_and_display(shape):
    print(f"area of {shape} is {shape.area()}")
    print(f"perimeter of {shape} is {shape.perimeter()}")


rectangle = Rectangle(10,20)
circle = Circle(10)

calculate_and_display(rectangle)
calculate_and_display(circle)
```

```python
class bankaccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance        # 🔒 private

    def __str__(self):
        return f"Account Owner: {self.owner} | Balance: ₹{self.__balance}"

    def get_balance(self):              # getter
        return self.__balance

    def deposit(self, deposit_amount):
        try:
            if not isinstance(deposit_amount, (int, float)):
                raise Exception("amount should be a number")
            if deposit_amount <= 0:
                raise Exception("amount should be positive")
            self.__balance += deposit_amount
            return f"Deposited ₹{deposit_amount}. Balance: ₹{self.__balance}"
        except Exception as e:
            return f"Error: {e}"

    def withdrawal(self, withdraw_amount):
        try:
            if not isinstance(withdraw_amount, (int, float)):
                raise Exception("amount should be a number")
            if withdraw_amount <= 0:
                raise Exception("amount should be positive")
            if withdraw_amount > self.__balance:
                raise Exception("insufficient balance")
            self.__balance -= withdraw_amount
            return f"Withdrawn ₹{withdraw_amount}. Balance: ₹{self.__balance}"
        except Exception as e:
            return f"Error: {e}"


class saving_account(bankaccount):
    def __init__(self, owner, balance=0, interest=0.025):
        super().__init__(owner, balance)
        self.__interest = interest      # 🔒 private

    def __str__(self):
        return (f"Savings Account | Owner: {self.owner} | "
                f"Balance: ₹{self.get_balance()} | "         # ✅ using getter
                f"Interest Rate: {self.__interest * 100}%")

    def get_interest(self):             # getter
        return self.__interest

    def set_interest(self, new_rate):   # setter with validation
        if not isinstance(new_rate, float):
            print("Interest rate must be a float")
        elif new_rate <= 0:
            print("Interest rate must be positive")
        else:
            self.__interest = new_rate
            print(f"Interest rate updated to {self.__interest * 100}%")

    def add_interest(self):
        interest_amount = self.get_balance() * self.__interest   # ✅ getter
        # no direct __balance access — using deposit() to add interest
        return self.deposit(interest_amount)


class current_account(bankaccount):
    def __init__(self, owner, balance=0, overdraft_limit=10000,
                 overdraft_interest=0.02, transaction_fee=10,
                 free_transaction_count_per_month=45, monthly_charge=0.02):
        super().__init__(owner, balance)
        self.__overdraft_limit = overdraft_limit            # 🔒 private
        self.__overdraft_interest = overdraft_interest      # 🔒 private
        self.__transaction_fee = transaction_fee            # 🔒 private
        self.__transaction_count = 0                        # 🔒 private
        self.__free_transactions = free_transaction_count_per_month  # 🔒 private
        self.__monthly_charge = monthly_charge              # 🔒 private

    def __str__(self):
        return (f"Current Account | Owner: {self.owner} | "
                f"Balance: ₹{self.get_balance()} | "        # ✅ getter
                f"Overdraft Limit: ₹{self.__overdraft_limit} | "
                f"Transactions: {self.__transaction_count}/{self.__free_transactions}")

    def get_overdraft_limit(self):        # getter
        return self.__overdraft_limit

    def get_transaction_count(self):      # getter
        return self.__transaction_count

    def deposit(self, deposit_amount):
        result = super().deposit(deposit_amount)
        if not result.startswith("Error"):
            self.__transaction_count += 1
            return f"Deposited ₹{deposit_amount}. Balance: ₹{self.get_balance()}"  # ✅ getter
        return result

    def withdrawal(self, withdraw_amount):
        try:
            if not isinstance(withdraw_amount, (int, float)):
                raise Exception("withdraw amount should be a number")
            if withdraw_amount <= 0:
                raise Exception("withdraw amount should be positive")
            if withdraw_amount > self.get_balance() + self.__overdraft_limit:  # ✅ getter
                raise Exception("exceeds overdraft limit")
            # use parent's internal logic via a workaround
            self._bankaccount__balance -= withdraw_amount   # accessing parent private
            self.__transaction_count += 1
            return f"Withdrawn ₹{withdraw_amount}. Balance: ₹{self.get_balance()}"
        except Exception as e:
            return f"Error: {e}"

    def monthly_charges(self):
        print(f"\n--- Monthly Charges for {self.owner} ---")

        monthly_charge_amount = self.get_balance() * self.__monthly_charge
        if self.get_balance() - monthly_charge_amount < -self.__overdraft_limit:
            print("Monthly charge exceeds overdraft limit!")
        else:
            self._bankaccount__balance -= monthly_charge_amount
            print(f"Monthly fee: ₹{monthly_charge_amount}. Balance: ₹{self.get_balance()}")

        if self.get_balance() < 0:
            overdraft_interest_amount = abs(self.get_balance()) * self.__overdraft_interest
            self._bankaccount__balance -= overdraft_interest_amount
            print(f"Overdraft interest: ₹{overdraft_interest_amount}. Balance: ₹{self.get_balance()}")
        else:
            print("No overdraft interest.")

        if self.__transaction_count > self.__free_transactions:
            extra = self.__transaction_count - self.__free_transactions
            charges = extra * self.__transaction_fee
            self._bankaccount__balance -= charges
            print(f"Extra transaction charges ({extra} transactions): ₹{charges}. Balance: ₹{self.get_balance()}")
        else:
            print(f"No extra transaction charges. ({self.__transaction_count}/{self.__free_transactions} used)")

        self.__transaction_count = 0
        print(f"--- Final Balance: ₹{self.get_balance()} ---\n")


# ✅ Objects
sai = saving_account("Saikiran", balance=5000)
print(sai)
print(sai.deposit(1000))
print(sai.add_interest())

ram = current_account("Ramesh", balance=15000)
print(ram)
print(ram.deposit(2000))
print(ram.withdrawal(5000))
ram.monthly_charges()
print(ram)

# ✅ Testing private protection
print(ram.get_balance())         # ✅ works via getter
print(ram.get_overdraft_limit()) # ✅ works via getter
# ram.__balance = -999999        # ❌ won't affect the real __balance
```

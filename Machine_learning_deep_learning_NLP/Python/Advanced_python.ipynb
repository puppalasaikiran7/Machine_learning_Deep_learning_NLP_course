```python
# __repr__ ( it is used for representing the object ) ( this is for developers ) 

#real world example 

from datetime import datetime
class Blogpost:
    def __init__(self,title,author,content,date):
        self.__title = title
        self.__author=author 
        self.__content = content 
        self.__date =date

    def __str__(self):
        return f"{self.__title} by {self.__author}"
    
    def __repr__(self):
        return f"Blogpost(title={self.__title}, author={self.__author}, content={self.__content},date={self.__date})"


post = Blogpost("my first blog is about how to learn ai the best way ","saikiran",
        "First learn the python from basic to advance and then go with data analytics and then machine learning from krish naik and then  gen ai from krish naik and then agentic ai from krish naik",
        datetime.now())


print(post)
print(repr(post))
```

## operator overloading 

### common operator overloading magic methods 
### __add__
### __sub__
### __eq__
### __lt__
### __gt__


```python
class Vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    
    def __add__(self,other):
        return Vector(self.x+other.x,self.y+other.y)

    def __sub__(self,other):
        return Vector(self.x-other.x,self.y-other.y)
    
    def __eq__(self,other):
        return self.x==other.x and self.y==other.y

    def __mul__(self,other):
        return Vector(self.x*other.x,self.y*other.y)
    
    def __repr__(self):
        return f"Vector({self.x},{self.y})"

v1=Vector(10,20)
v2=Vector(20,30)

print(v1+v2)
print(v1-v2)
print(v1==v2)
print(v1*v2)
print(repr(v1))
print(repr(v2))
```

```python
#inbuild exception 

class banking:
    def __init__(self,balance):
        self.balance = balance 
    
    def withdraw(self,amount):
        if amount >self.balance:
            raise ValueError("not enough money ")
        if amount <0:
            raise ValueError("nagative amount")
        return self.balance-amount
    


Canarabank = banking(1000)
try:
    print(Canarabank.withdraw(100))
except ValueError as e:
    print(e)


Kotakbank = banking(900)
try:
    print(Kotakbank.withdraw(1000))
except ValueError as e:
    print(e)
```

```python
#custom exception 

#basic exception
class InsufficientFundsError(Exception):
    pass

class NegativeAmountError(Exception):
    pass

class banking:
    def __init__(self,balance=0):
        self.balance = balance 
    
    def withdraw(self,amount):
        if amount <0:
            raise NegativeAmountError("amount cannot be negative")
        if amount>self.balance:
            raise InsufficientFundsError("insufficient funds")
        return self.balance-amount

    def deposite(self,amount):
        if amount<0:
            raise NegativeAmountError("cannot deposite negative amount")
        self.balance+=amount
        return f"updated balance is :{self.balance}"

    def __str__(self):
        return f"balance: {self.balance}"


bwj = banking(1000)
print(bwj)
try:
    bwj.withdraw(-5000)
except InsufficientFundsError as e:
    print(e)
except NegativeAmountError as e:
    print(e)
except Exception as e:
    print(e)


try:
    print(bwj.deposite(4000))
except NegativeAmountError as e:
    print(e)
```

```python
## custom exception with message 
# in this we are not using the super().__init__(message) because we are using the __str__ , 
# if we are not defining
# __str__ then we need to  use the super().__init__(message)


class AgeError(Exception):
    def __init__(self,age,message):
        self.age = age 
        self.message =message 
    
    def __str__(self):
        return f"{self.message} :{self.age}"
        

age = -5 
try:
    raise AgeError(age,"age cannot be negative")
except AgeError as e:
    print(e)
    
```

```python
def testing():
    yield "this is cup1"
    yield "this is cup2"
    yield "this is cup3"

test_case = testing()
print(test_case)   # this is basically giving the location to the testing function and the data  

for i in test_case:
    print(i)
```

Generators are used for reading large files because they allow you to process one line at a time with out loading the entire file into memory

```python
def read_large_file(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line 

file_path = 'myfile.txt'
 
for line in read_large_file(file_path):
    print(line.strip())
```

```python
## decorator 

def main_welcome(func):
    def wrapper():
        print("welcome to the main function")
        func
    return wrapper

@main_welcome
def main():
    print("this is the main function")

main()
```

```python
##real example 

class User:
    def __init__(self,name,is_authenticated):
        self.name = name 
        self.is_authenticated=is_authenticated

def require_auth(func):
    def wrapper(user,*args,**kwargs):
        print(f"cheking if {user.name} is authenticated")
        if user.is_authenticated:
            print(f"{user.name} is authenticated proceed")
            return func(user,*args,**kwargs)
        else:
            return f"{user.name} is not authenticated"
    return wrapper

@require_auth
def view_dashboard(user):
    return f"welcome to dashboard {user.name}"



user1= User("Alice",True)
user2= User("Bob",False)


print(view_dashboard(user1))
print(view_dashboard(user2))
```
